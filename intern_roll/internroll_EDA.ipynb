```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_json('roll.json')
df.head()
```

```python
df.users.iloc[0]
```

```python
df = pd.DataFrame(df.users.to_list())
df.head()
```

```python
len(df)
```

```python
def get_course_name(row):
    if row.enrolledCourses:
        return row.enrolledCourses[0].get('courseName')
    else:
        return None
```

```python
df['course_name'] = df.apply(get_course_name, axis=1)
```

```python
enrolled_df = df[df['course_name'].notnull()]
enrolled_df.head()
```

```python
len(enrolled_df)
```

```python
enrolled_df.course_name.value_counts()
```

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(enrolled_df.course_name.value_counts(), orient='h',)
ax.bar_label(ax.containers[0], fontsize=10, padding=5);
```

```python
enrolled_df.course_name.value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(8, 6))
```

```python
enrolled_df.to_csv('enrolled.csv')
```
