```python
class Node:
    def __init__(self, data):
        self.val = data
        self.left = None
        self.right = None
        
class BinaryTree:
    def __init__(self):
        self.root = None

    def __r_insert(self,current_node,value):
        if current_node == None:
            self.root = Node(value)
            return Node(value)
        elif current_node.val > value:
            if current_node.left is not None:
                self.__r_insert(current_node.left,value)
            else:
                current_node.left = Node(value)
                return current_node.left
        elif current_node.val < value:
            if current_node.right is not None:
                self.__r_insert(current_node.right,value)
            else:
                current_node.right = Node(value)
                return current_node.right
        elif current_node.val == value:
            return current_node
    
    def r_insert(self,value):
        return self.__r_insert(self.root,value)
        
        
```

```python
tree = BinaryTree()
tree.r_insert(10)
tree.r_insert(5)
tree.r_insert(75)
tree.r_insert(325)
tree.r_insert(1)
tree.r_insert(1125)
```

```python
def BFS(root: Node):
    q = [root]
    r = []
    current_node = root
    while len(q) > 0:
        # print(current_node.val)
        if current_node.left != None:
            q.append(current_node.left)
        if current_node.right != None:
            q.append(current_node.right)
        r.append(current_node.val)
        q.pop(0)
        if len(q) > 0:
            current_node = q[0]
            
    return r
        
BFS(tree.root)
        
```

```python
def DFS_preorder(root: Node):
    r = []
    def r_dfs(current_node):
        r.append(current_node.val)
        if current_node.left is not None:
            r_dfs(current_node.left)
        if current_node.right is not None:
            r_dfs(current_node.right)
    r_dfs(root)
    return r
DFS_preorder(tree.root)
```

```python
def DFS_postorder(root: Node):
    r = []
    def r_dfs(current_node):
        if current_node.left is not None:
            r_dfs(current_node.left)
        if current_node.right is not None:
            r_dfs(current_node.right)
        r.append(current_node.val)
        
    r_dfs(root)
    return r
DFS_postorder(tree.root)
```

```python
def DFS_inorder(root: Node):
    result = []
    def r_dfs(current_node: Node):
        if current_node.left is not None:
            r_dfs(current_node.left)
        result.append(current_node.val)
        if current_node.right is not None:
            r_dfs(current_node.right)
    r_dfs(root)
    
    return result
DFS_inorder(tree.root)
```
