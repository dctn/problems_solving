```python
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-1):
            if arr[j] < arr[j+1]:
                temp = arr[j]
                arr[j] = arr[j+1]
                arr[j+1] = temp
    return arr
```

```python
print(bubble_sort([5,4,3,2,1]))
```

# Selection Sort

```python
def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(len(arr)):
            if arr[min_idx] > arr[j]:
                min_idx = j
        temp = arr[i]
        arr[i] = arr[min_idx]
        arr[min_idx] = temp        
    return arr
```

```python
print(selection_sort([5,4,3,2,1]))
```
