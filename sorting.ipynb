```python
# bubble sorting algorithm own code
def bubble_sort(sort_list):
    need_swap = True
    end = len(sort_list) -1
    
    while need_swap:
        need_swap = False
        for i in range(end):
            if sort_list[i] > sort_list[i+1]:
                # print(sort_list)
                temp = sort_list[i]
                sort_list[i] = sort_list[i+1]
                sort_list[i+1] = temp
                need_swap=True
        end -=1
        
    return sort_list
    
    
    
print(bubble_sort([1,5,5,2,6,0]))
```

# Selection Sort

```python
def selection_sort(arr):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i, len(arr)):
            if arr[min_idx] > arr[j]:
                min_idx = j
        temp = arr[i]
        arr[i] = arr[min_idx]
        arr[min_idx] = temp        
    return arr
print(selection_sort([5,4,3,2,1]))
```

```python
# own code for selection sort
def selection_sort(sort_arr):
    
    for i in range(len(sort_arr)):
        min_index = i
        for num in range(i,len(sort_arr)):
            if sort_arr[num] < sort_arr[min_index]:
                min_index = num
            
        temp = sort_arr[i]
        sort_arr[i] = sort_arr[min_index]
        sort_arr[min_index] = temp
    return sort_arr
            
    
    
    
print(selection_sort([1,5,2,6,0]))
```

# insertion sort

```python
# own code for insertion sort
def insertion_sort(sort_list):
    
    for i in range(1,len(sort_list)):
        # move_idx = i
        for idx in range(i,0,-1):
            if sort_list[idx] < sort_list[idx-1]:
                temp = sort_list[idx]
                sort_list[idx] = sort_list[idx-1]
                sort_list[idx-1] = temp
            elif sort_list[idx] > sort_list[idx-1]:
                continue
        
    return sort_list
            
        
        
print(insertion_sort([3,5,2,0,4]))
```
