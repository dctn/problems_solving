```python
def reverse_str(str_):
    return str_[::-1]

reverse_str('python')
```

```python
def check_palinrome(str_):
    left = 0
    right = len(str_)-1
    
    while left < right:
        if str_[left] == str_[right]:
            left +=1
            right -=1
        else:
            return False
    return True

check_palinrome('madam0')
```

```python
def factorial(n):
    fac = 1
    for i in range(1,n+1):
        fac *= i
        
    return fac
factorial(5)
```

```python
def feb(n):
    prev = 0
    cur = 1
    print(prev,cur,end=" ")
    
    for i in range(n-2):
        sum_ = prev + cur
        print(sum_,end=" ")
        prev = cur
        cur = sum_

feb(7)
```

```python
def group_anagrams(arr):
    
    my_dict = {}
    
    for string in arr:
        sort = ''.join(sorted(string))
        
        if sort in my_dict:
            my_dict[sort].append(string)
        else:
            my_dict[sort] = [string]
    print(list(my_dict.values()))
    return list(my_dict)
group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
```
