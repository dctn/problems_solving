```python
class VariableManager:
    def __init__(self):
        self.global_variables = {}
        self.transactions = []
        self.cmd_shell()
        
    def begin(self):
        self.transactions.append({})
        return self.transactions[-1]
    
    def get(self,key):    
        for i in range(len(self.transactions + [self.global_variables])-1,-1,-1):
            if i == len(self.transactions):
                transaction = self.global_variables
            else:
                transaction = self.transactions[i]
                
            if key in transaction:
                return transaction[key]
        return None
    
    def roll_back(self):
        self.transactions.pop()
        if len(self.transactions) > 0:
            return self.transactions[-1]
        else:
            return self.begin()
    
    def commit(self):
        for key in self.transactions[-1]:
            self.global_variables[key] = self.transactions[-1][key]
        self.transactions = []
    
    def cmd_shell(self):
        is_exit = False
        cur_transacation = self.begin()
        while not is_exit:
            input_str = input("Enter a number (or 'exit' to quit): ").lower()
            input_list = input_str.split(" ")
            cmd = input_list[0]
            if cmd == 'exit':
                is_exit = True
                
            elif cmd == 'set':
                cur_transacation[input_list[1]] = input_list[2]
                # if input_list[2] not in self.val_count:
                #     self.val_count[input_list[2]] = 1
                # else:
                #     self.val_count[input_list[2]] += 1
                
            elif cmd == 'get':
                print(self.get(input_list[1]))
                    
            elif cmd == 'unset':
                if input_list[1] in self.global_variables:
                    del cur_transacation[input_list[1]]
                
                else:
                    print(None)
                    
            elif cmd == 'count':
                print(self.count(input_list[1]))
                
            elif cmd == 'begin':
                cur_transacation = self.begin()
            elif cmd == 'rollback':
                cur_transacation = self.roll_back()
            elif cmd == 'commit':
                self.commit()
                
    def count(self,val):
            count = 0
            seen = set()
            for i in range(len(self.transactions + [self.global_variables])-1,-1,-1):
                transacation = self.global_variables if i == len(self.transactions) else self.transactions[i]
                # print(transacation)
                for key in transacation:
                    if key not in seen and transacation[key] == val:
                        count +=1
                    seen.add(key)
                # print(seen)
                
            return count
```
