```python
def print_num(n):
    if n==0:
        return 0
    print(n)
    print_num(n - 1)
```

```python
def sum_n(n):
    if n == 1:
        return 1
    small_ans = sum_n(n-1)
    ans = n + small_ans
    return ans
sum_n(5)
```

```python
def print_reversed(n):
    if n == 1:
        print(1)
        return
    print(n)
    print_reversed(n-1)
    
print_reversed(3)
```

```python
def fac(n):
    if n==1:
        return 1
    small_ans = fac(n-1)
    ans = small_ans * n
    return ans

fac(5)
```

```python
def sum_digits(n):
    if n <= 9:
        return n
    
    last = n % 10
    rem = n// 10
    small_ans = sum_digits(rem)
    ans = small_ans + last
    return ans

sum_digits(911)
```

```python
def no_of_digit(n):
    if n <= 9:
        return 1
    
    last = n % 10
    rem = n//10
    small_ans = no_of_digit(rem)
    ans = small_ans + 1
    return ans

no_of_digit(0)
```

```python
def count_zero(n):
    if n == 0:
        return 1
    
    last = n%10
    rem = n // 10
    
    small_ans = count_zero(rem)
    if last == 0:
        ans = 1 + small_ans
    elif last != 0 :
        ans = 0 + small_ans
    return ans

count_zero(101001)
```

```python
def remove_zero(n):
    if n  == 0:
        return 0
    
    last = n % 10
    rem = n // 10
    small_ans = remove_zero(rem)
    ans = (small_ans * 10) + last if last != 0 else small_ans
    return ans

remove_zero(1005)
```

```python
def print_sum(n):
    if n <= 9:
        print(n)
        return n
    
    last = n%10
    rem = n//10
    
    small_num = print_sum(rem)
    ans = last + small_num
    print(ans)
    return ans

print_sum(12)
```

```python
def reverse(n,rev=0):
    if n==0:
        return rev
    last = n%10
    rem = n//10
    rev = (rev * 10) + last
    small_ans = reverse(rem,rev)
    return small_ans

reverse(123)
```

```python
def reverse_even(n,rev=0):
    if n == 0:
        return rev
    last = n %10
    rem = n//10
    if last %2 ==0:
        rev = (rev *10) + last
    small_ans = reverse_even(rem,rev)
    return small_ans

reverse_even(2254)
```

```python
def sum_odd(n):
    if n <=9:
        return n
    
    rem = n // 10
    last = n%10
    small_ans = sum_odd(rem)
    if last %2 ==1:
        ans = last + small_ans
    else:
        ans = small_ans
    return ans

sum_odd(12345)
```

```python
def digit_product(n):
    if n <=9:
        return n
    
    last = n%10
    rem = n//10
    small_ans = digit_product(rem)
    ans = last * small_ans
    return ans

digit_product(121)
```

```python
def check_all_digit_even(n):
    if n<=9 :
        return True if n%2==0 else False
    last = n%10
    rem = n//10
    small_ans = check_all_digit_even(rem)
    if last%2== 0 and small_ans:
        ans = True
    else:
        ans= False
    return ans

check_all_digit_even(222)
```

```python
def sum_even(n):
    if n  == 0:
        return n

    
    small_ans = sum_even(n-1)

    ans = n+small_ans if n%2 ==0 else small_ans
    return ans

sum_even(4)
```

```python
def sum_multiple_3(n):
    if n == 0:
        return n
    small_ans = sum_multiple_3(n-1)
    ans = small_ans+n if n%3 == 0 else small_ans
    return ans

sum_multiple_3(10)
```

```python
def sum_from_a_b(a,b):
    if a==b:
        return a
    small_ans = sum_from_a_b(a+1,b)
    ans = a + small_ans
    return ans
sum_from_a_b(1,3)
```

```python
def sum_even_from_a_b(a,b):
    if a==b:
        return a if a%2 ==0 else 0
    small_ans = sum_even_from_a_b(a+1,b)
    ans = a+small_ans if a%2==0 else small_ans
    return ans

sum_even_from_a_b(1,5)
```

```python
def power_of_x(x,n):
    if n==0:
        return 1
    
    small_ans = power_of_x(x,n-1)
    ans = n * small_ans
    return ans

power_of_x(2,3)
```

```python
def reverse(n,rev=0):
    if n==0:
        return rev
    last = n%10
    rem = n//10
    rev = (rev*10) + last
    ans = reverse(rem,rev)
    return ans
reverse(123)
```

```python
def sum_arr(arr):
    if len(arr) == 0:
        return 0
    elif len(arr) == 1:
        return arr[0]
    
    small_ans = sum_arr(arr[1:])
    ans = arr[0] + small_ans
    return ans

print(sum_arr([1,2,3]))
```
