```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class Stack:
    def __init__(self,value):
        new_node = Node(value)
        self.top = new_node
        self.hieght = 1
    def push(self,value):
        new_node = Node(value)
        if self.top is None:
            self.top = new_node
        else:
            new_node.next = self.top
            self.top = new_node
        self.hieght += 1
        
    def pop(self):
        if self.top is None:
            return None
        elif self.top.next is None:
            self.top = None
            self.hieght -= 1
            return None

        temp = self.top
        self.top = self.top.next
        temp.next = None
        self.hieght -= 1
        return temp
    
    def print_stack(self):
        temp = self.top
        while temp != None:
            print(temp.value)
            temp = temp.next
```

```python
stack = Stack(2)
stack.push(3)
stack.print_stack()
```

```python
stack.pop()
stack.print_stack()
```
