# max sum of subarray with k size

```python
# normal version of the sliding window technique

def max_sub_array_k_size(arr,k):
    max_sum = float('-inf')
    for i in range(len(arr)-k+1):
        current_sum = sum(arr[i:i+k])
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum
max_sub_array_k_size([12,3,5,6,7,1],3)
```

```python
# normal version of the sliding window technique for leetcode problem : longest sum of subarray without repeatation 
def max_sub_array_k_size(arr,k):
    max_sum = float('-inf')
    arr_start = 0
    for i in range(0,(len(arr)-k)):
        curren_sum = sum(arr[i:(i+k)-1])
        if curren_sum > max_sum and len(set(arr[i:(i+k)-1])) == len(arr[i:(i+k)-1]):
            max_sum = curren_sum
            arr_start = i
    return max_sum,arr[arr_start:arr_start+k]

max_sub_array_k_size([2,3,5,7,6,1],3)
```

```python
# optimal version of the sliding window technique for leetcode problem : longest sum of subarray without repeatation 

def max_sub_array_k_size(arr,k):
    max_sum = float('-inf')
    for i in range(len(arr) - k+1):
        if i == 0:
            current_sum = sum(arr[i:k])
        else:
            current_sum -= arr[i-1]
            current_sum += arr[(i+k)-1]
        
        if current_sum > max_sum and len(set(arr[i:(i+k)])) == len(arr[i:(i+k)]):
            max_sum = current_sum
    return max_sum            
# max_sub_array_k_size([12,3,5,6,7,1],3)
max_sub_array_k_size([1,5,4,2,9,9,9],3)

```

# max product of subarray

```python
def max_product_of_subarray_k_size(arr,k):
    max_product = float('-inf')
    current_product = 1
    max_index = 0
    for i in range(len(arr)-k+1):
        if i==0 :
            for num in arr[i:i+k]:
                current_product *= num
        else:
            current_product /= arr[i-1]
            current_product *= arr[(i+k)-1]
            
        if current_product > max_product:
            max_product = current_product
            max_index = i
        # current_product = 1
    return max_product, max_index

max_product_of_subarray_k_size([4, 2, 1, -9, 8, 2, 3], 3)
```

# subarray with target sum value

```python
def subarray_target_sum_size_k(nums, target, k):
  count = 0

  for i in range(len(nums)-k+1):
    if i == 0:
      current_pro = sum(nums[i:i+k])
    else:
      current_pro -= nums[i-1]
      current_pro += nums[(i+k)-1]

    if current_pro == target:
      count +=1

  return count
```

# is string contain anagram

```python
def is_string_has_anagram(string,sub_string):
    sub_string_list = list(sub_string)
    is_same = 0
    
    for i in range(len(string)-len(sub_string)+1):
        current_str = string[i:i+len(sub_string)]
        
        for j in range(len(sub_string_list)):
            if current_str[j] in sub_string_list:
                is_same +=1
                
        if is_same == len(sub_string_list):        
            return True
        is_same = 0
    return False
is_string_has_anagram('suyma','ma')
```

# count substring anagrams

```python
from collections import Counter

def anagram_count(string,sub_string):
    sub_string_map = Counter(sub_string)
    anagram_count_value = 0
    
            
    # sliding window with mapping of anagram count
    for i in range(len(string)-len(sub_string)+1):
        
        if i == 0:
            current_string_map = Counter(string[i:len(sub_string)])
        else:
            current_string_map[string[i-1]] -= 1
            current_string_map[string[(i+len(sub_string))-1]] += 1    
        
        
        if current_string_map == sub_string_map:
            anagram_count_value += 1
            
    return anagram_count_value
anagram_count("tacoctacabcatt", "cat")
```

# longest subarray sum

Write a function that takes in an list and a target sum. The function should return the length of the longest subarray that sums to the target.

You can assume that the elements of the list are nonnegative.

If there is no subarray that sums to the target, then return -1.



```python
def longest_sub_array_sum(arr,target):
    start = 0
    window_sum = 0
    arr_len = 0
    for end in range(0, len(arr)):
        window_sum += arr[end]
        
        while window_sum > target:
            window_sum -= arr[start]
            start +=1
            
        if target == window_sum:
            current_len = (end - start)+1
            if current_len > arr_len:
                arr_len = current_len
    return arr_len

longest_sub_array_sum([1, 2, 1, 5, 2, 3, 10, 1, 9, 4, 3, 3, 7], 10)
```

# longest unique substring
Write a function that takes in a string as an argument. The function should return the length of the longest substring that consists of only unique characters.

```python
def longest_distinct_substring_len(string):
    start = 0
    longest_len = 0
    window_counter  = []
    for end in range(len(string)):
        if end == 0:
            window_counter = string[end]
        else:
            window_counter = string[start:end+1]
        
        while len(list(window_counter)) != len(set(window_counter)):
            window_counter = window_counter[start+1:end+1]
            start += 1
            
        if len(list(window_counter)) == len(set(window_counter)):
            if longest_len <len(window_counter):
                longest_len = len(window_counter)
            
    return longest_len
        
longest_distinct_substring_len('abcabcqbb')
```

# Max of min for every window size

You are given an integer array arr[], the task is to find the maximum of minimum values for every window size k where 1≤ k ≤ arr.size().

For each window size k, consider all contiguous subarrays of length k, determine the minimum element in each subarray, and then take the maximum among all these minimums.

Return the results as an array, where the element at index i represents the answer for window size i+1.

```python
# brute force solution for the leetcode problem : maximum of minimum for every window size in an array
def max_of_min(arr):
    max_min = []
    for k in range(1,len(arr)+1):
        min_list = []
        for i in range(len(arr)-k+1):
            window_min = min(arr[i:i+k])
            min_list.append(window_min)
        
        max_min.append(max(min_list))
    
    return max_min
max_of_min([10, 20, 30])
```
